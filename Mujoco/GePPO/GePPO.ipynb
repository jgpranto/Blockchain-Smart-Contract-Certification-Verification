### The Theoretical Change (On-Policy PPO vs. Off-Policy GePPO)
Standard PPO is strictly **on-policy**: it centers its clipping range at $1 \pm \epsilon$, which forces it to discard all data after a single update.

GePPO extends PPO to the **off-policy** setting by reusing samples from the last $M$ prior policies ($\pi_{k-i}$).
* **PPO Clip:** `torch.clamp(ratio, 1.0 - eps, 1.0 + eps)`
* **GePPO Clip:** `torch.clamp(ratio, ref_ratio - eps, ref_ratio + eps)`
